# R-GRAV-001: Pemodelan Sintetik Anomali Gravitasi akibat Intrusi Bawah Permukaan

## 1. IDEA — Identifikasi Masalah

### Latar Belakang

Metode gravitasi merupakan metode geofisika yang digunakan untuk mempelajari variasi densitas batuan bawah permukaan melalui perubahan medan gravitasi. Intrusi batuan yang memiliki densitas berbeda dari batuan sekitarnya dapat menghasilkan anomali gravitasi.

Pada proyek ini, dilakukan pemodelan sintetik anomali gravitasi akibat intrusi batuan berbentuk bola homogen dengan kontras densitas positif. Model dibuat menggunakan parameter teoritis dan tidak menggunakan data pengukuran lapangan.

### Tujuan

1. Membuat model sintetik intrusi batuan bawah permukaan.
2. Menghitung respons gravitasi teoritis akibat intrusi.
3. Memvisualisasikan model geologi dan kurva anomali gravitasi.
4. Menganalisis pengaruh kedalaman, ukuran, dan kontras densitas terhadap respons gravitasi.
5. Menghasilkan notebook Python yang dapat dijalankan dan dikembangkan kembali.

### Pertanyaan Utama

Bagaimana pengaruh kedalaman, ukuran, dan kontras densitas intrusi terhadap amplitudo dan bentuk anomali gravitasi di permukaan?

## 2. CLARIFY — Klarifikasi Kebutuhan

### Kebutuhan

Program harus mampu menerima parameter model, menghitung respons gravitasi teoritis, menampilkan model sintetik, dan menghasilkan kurva anomali gravitasi sepanjang lintasan pengamatan.

### Asumsi Model

- Intrusi berbentuk bola homogen.
- Permukaan bumi dianggap datar.
- Kontras densitas intrusi terhadap batuan sekitar bernilai positif.
- Pengamatan dilakukan sepanjang lintasan horizontal di permukaan.
- Perhitungan menggunakan pendekatan gravitasi Newton.
- Seluruh parameter merupakan parameter sintetik.

### Batasan

Model tidak memperhitungkan topografi, variasi densitas internal, sumber anomali lain, koreksi gravitasi lapangan, maupun noise pengukuran.

Oleh karena itu, hasil pemodelan digunakan untuk pembelajaran dan pengujian metode *forward modeling*, bukan sebagai interpretasi langsung kondisi geologi sebenarnya.

## 5. IMPLEMENT — Implementasi Model

### Dasar Teori

Volume intrusi berbentuk bola dihitung dengan persamaan:

$$
V = \frac{4}{3}\pi r^3
$$

Anomali gravitasi vertikal pada titik pengamatan di permukaan dihitung menggunakan persamaan:

$$
\Delta g(x) =
\frac{G\Delta\rho Vz_0}
{\left[(x-x_0)^2+z_0^2\right]^{3/2}}
$$

Keterangan:

- G adalah konstanta gravitasi universal.
- Δρ adalah kontras densitas batuan.
- V adalah volume intrusi.
- z0 adalah kedalaman pusat intrusi.
- x0 adalah posisi horizontal pusat intrusi.
- x adalah posisi horizontal titik pengamatan.

Konversi satuan dari m/s² ke mGal:

$$
\Delta g_{\mathrm{mGal}}
=
\Delta g_{\mathrm{m/s^2}}\times10^5
$$
